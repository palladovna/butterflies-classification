```python
import torch

from src.data.dataloader import get_dataloaders
from src.models.myalexnet import create_custom_alexnet
from src.models.resnet import create_resnet18
from src.models.vgg import create_vgg16
from src.utils.evaluate import evaluate

import pandas as pd
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
train_loader, val_loader, test_loader = get_dataloaders(
    train_dir=r"C:\Users\MSI\Desktop\project\Butterflies\train",
    val_dir=r"C:\Users\MSI\Desktop\project\Butterflies\val",
    test_dir=r"C:\Users\MSI\Desktop\project\Butterflies\test",
    batch_size=32,
    image_size=224,
    num_workers=0
)
```

```python
def evaluate_model(model_fn, weights_path, num_classes=100):
    """
    model_fn — функция создания модели (create_resnet18, create_vgg16...)
    weights_path — путь к весам
    """
    # Создаём модель
    model = model_fn(num_classes=num_classes)

    # Загружаем веса
    model.load_state_dict(torch.load(weights_path, map_location=device))

    # Переносим на GPU/CPU
    model.to(device)

    # Вызываем готовую evaluate()
    return evaluate(model, test_loader, device)
```

```python
alex_metrics = evaluate_model(
    create_custom_alexnet,
    "C:/Users/MSI/Desktop/project/butterflies-classification/weights/alexnet_custom_best.pth",
    num_classes=100
)
alex_metrics
```

```python
resnet18_metrics = evaluate_model(
    create_resnet18,
    "C:/Users/MSI/Desktop/project/butterflies-classification/weights/resnet18_best.pth",
    num_classes=100
)
resnet18_metrics
```

```python
vgg16_metrics = evaluate_model(
    create_vgg16,
    "C:/Users/MSI/Desktop/project/butterflies-classification/weights/vgg16_best.pth",
    num_classes=100
)
vgg16_metrics
```
